## Completion Model

```python
from langchain_google_genai import GoogleGenerativeAI

llmModel = GoogleGenerativeAI(model="gemini-3.5-flash-lite", temperature=0.7)
```

## Chat Model

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llmChatModel = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite", temperature=0.7)
```

## Prompts and prompts template

```python
# For Completion Model
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate.from_template("Write a {adjective} story about the {topic}")

prompt = prompt_template.format(adjective="funny", topic="cat")

response = llmModel.invoke(prompt)
```

```python
response
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a {profession} expert on {topic}"},
    {"role": "user", "content": "Hello Mr {profession}, Can you help me with {topic}"},
    {"role": "assistant", "content": "Sure"},
    {"role": "user", "content": "{input}"}
])

messages = chat_prompt_template.format_messages(profession="doctor", topic="cardiology", input="I have chest pain.")

response = llmModel.invoke(messages)
print(response)
```

## Older Prompt Template Method

```python
from langchain_core.prompts import HumanMessagePromptTemplate
from langchain_core.messages import SystemMessage

chat_templates = ChatPromptTemplate.from_messages([
    SystemMessage(content="You are a {profession} expert on {topic}"),
    HumanMessagePromptTemplate.from_template("{input}"),
])

messages = chat_templates.format_messages(input="I have chest pain.", profession="doctor", topic="cardiology")

response = llmModel.invoke(messages)
print(response)
```

## Few Shot Prompting

```python
from langchain_core.prompts import FewShotChatMessagePromptTemplate
examples = [
    {"input": "Hello", "output": "Hi there! How can I help you?"},
    {"input": "I have a headache.", "output": "You should see a doctor for a proper diagnosis."}
]

examples_prompt = ChatPromptTemplate.from_messages(
    [
        {"role": "user", "content": "{input}"},
        {"role": "assistant", "content": "{output}"}
    ]
)

few_shot_prompt = FewShotChatMessagePromptTemplate(
    examples=examples,
    example_prompt=examples_prompt
)

final_prompt = ChatPromptTemplate.from_messages(
    [
        few_shot_prompt,
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "{input}"}
    ]
)
```

```python
response = llmModel.invoke(final_prompt.format_messages(input="I have a headache."))
print(response)
```
