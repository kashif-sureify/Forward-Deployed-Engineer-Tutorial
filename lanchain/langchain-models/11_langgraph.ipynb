```python
import os
from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv())
gemini_api_key = os.getenv("GEMINI_API_KEY")
gemini_model = os.getenv("GEMINI_MODEL")
groq_api_key = os.getenv("GROQ_API_KEY")
tavily_api_key = os.getenv("TAVILY_API_KEY")
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import GoogleGenerativeAI, ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
from langchain_community.tools.tavily_search import TavilySearchResults

search = TavilySearchResults(api_key=tavily_api_key, max_results=3)

model = ChatGoogleGenerativeAI(api_key=gemini_api_key, model=gemini_model)
parser = StrOutputParser()
```

```python
response = search.invoke("What is the current weather in New York?")
print(response)
```

```python
tools = [search]
tools
```

```python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage

agent = create_agent(model=model, tools=tools)
```

```python
response = agent.invoke({"messages": [HumanMessage(content="What is the current weather in New York?")]})
print(response)
```

## Adding memory
Adding memory in LangGraph is very similar to what we did with LangChain.

```python
from langgraph.checkpoint.memory import MemorySaver

memory_saver = MemorySaver()
```

```python
agent = create_agent(model=model, tools=tools, checkpointer=memory_saver)
```

```python
config = {
    "configurable" :{
        "thread_id": "001"
    }
}
```

```python
for chunk in agent.invoke({"messages": [HumanMessage(content="Tell me about top 5 news of today")]}, config=config).get("messages", []):
    print(chunk)
    print("------")
```

```python
for chunk in agent.invoke({"messages": [HumanMessage(content="Who are Houthis")]}, config=config).get("messages", []):
    print(chunk)
    print("------")
```

```python
config2 = {"configurable": {"thread_id": "002"}}
for chunk in agent.invoke({"messages": [HumanMessage(content="What are we talking about")]}, config=config2).get("messages", []):
    print(chunk)
    print("------")
```
