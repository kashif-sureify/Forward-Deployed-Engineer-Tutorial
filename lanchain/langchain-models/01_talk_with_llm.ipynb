```python
import os
from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv())
gemini_api_key = os.environ['GEMINI_API_KEY']
gemini_model = os.environ['GEMINI_MODEL']
```

# Completion Model

```python
# Completion Model Initialization
from langchain_google_genai import GoogleGenerativeAI

llm = GoogleGenerativeAI(api_key=gemini_api_key, model=gemini_model)

print(llm)
```

```python
response = llm.invoke("Who is the president of the United States?")
```

```python
print(response)
```

```python
for chunk in llm.stream("Tell me 200 words about the president of the United States"):
    print(chunk, end="", flush=True)
```

```python
# Temperature: It defines the randomness of the model's output. Higher values (e.g., 0.8) make the output more diverse, while lower values (e.g., 0.2) make it more focused and deterministic.

creativeLLM = GoogleGenerativeAI(api_key=gemini_api_key, model=gemini_model, temperature=0.8)
```

```python
response = creativeLLM.invoke("Tell me 200 words about the president of the United States")
print(response)
```

# Chat Model

```python
# Chat Model Initialization
from langchain_google_genai import ChatGoogleGenerativeAI

chat_model = ChatGoogleGenerativeAI(api_key=gemini_api_key, model=gemini_model, temperature=0.7)

print(chat_model)
```

# Old Method

```python
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="My name is Kashif."),
]

response = chat_model.invoke(messages)
```

```python
response
```

# New Method

```python
messages = [
    ("system", "You are a helpful assistant."),
    ("user", "My name is Kashif.")
]

response = chat_model.invoke(messages)
response.content[0]['text']
```

```python
response.schema()
```

```python
response.content[0]['text']
```

```python
from langchain_core.messages import AIMessage, HumanMessage
chat_history = []

def chat(message):
    chat_history.append(HumanMessage(content=message))
    response = chat_model.invoke(chat_history)
    chat_history.append(response)
    return response.text

print(chat("My name is Kashif."))
print(chat("What is my name?"))
```
