```python
from langchain_google_genai import GoogleGenerativeAI

model = GoogleGenerativeAI(model=gemini_model, api_key=gemini_api_key, temperature=0.3, timeout=30, max_retries=1)
```

```python
prompt = """You are a teacher creating practice questions about Data Structures from a textbook.

Text:
__________
{text}
__________

Write exactly 2 simple, clear questions that can be answered using the text above.

Rules:
- One question per line.
- Each question is one short sentence ending with a question mark.
- No numbering, headings, bullet points, tables, markdown, sub-questions, or any other text.

QUESTIONS:
"""
```

```python
from langchain_core.prompts import PromptTemplate

PROMPT_QUESTIONS = PromptTemplate(
    input_variables=["text"],
    template=prompt
)
```

```python
refine_template = """You are a teacher creating practice questions about Data Structures from a textbook.

Current questions:
{existing_answer}

More text from the textbook:
__________
{text}
__________

Improve the current questions using the new text if it helps. If it does not help, keep them unchanged.

Rules:
- Exactly 2 simple, clear questions, one question per line.
- Each question is one short sentence ending with a question mark.
- No numbering, headings, bullet points, tables, markdown, sub-questions, or any other text.

QUESTIONS:
"""
```

```python
REFINE_PROMPT = PromptTemplate(
    input_variables=["existing_answer", "text"],
    template=refine_template
)
```

```python
from langchain_classic.chains.summarize import load_summarize_chain
from langchain_groq import ChatGroq

model = ChatGroq(model="openai/gpt-oss-20b", api_key=groq_api_key, timeout=30, max_retries=3)

```

```python
question_gen_chain = load_summarize_chain(
    llm=model,
    chain_type="refine",
    verbose=True,
    question_prompt=PROMPT_QUESTIONS,
    refine_prompt=REFINE_PROMPT
)
```

```python
questions = question_gen_chain.invoke(document_question_gen)
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2")
```

```python
from langchain_chroma import Chroma

vectorstore = Chroma.from_documents(pages, embedding=embeddings, persist_directory="chroma_db")
```

```python
llm_ans_gen = GoogleGenerativeAI(model=gemini_model, api_key=gemini_api_key, temperature=0.3, timeout=30, max_retries=1)
```

```python
questions_text = questions["output_text"]

# keep only real questions (lines ending with "?")
questions_list = [q.strip() for q in questions_text.split("\n") if q.strip().endswith("?")]

print(len(questions_list), "questions")
for i, q in enumerate(questions_list, 1):
    print(f"{i}. {q}")
```

```python
from langchain_classic.chains import RetrievalQA
```

```python
answer_generation_chain = RetrievalQA.from_chain_type(
    llm=llm_ans_gen,
    chain_type="stuff",
    retriever=vectorstore.as_retriever(),
)
```

```python
with open("answers.txt", "w") as f:   # "w" = start fresh each run
    for i, q in enumerate(questions_list, 1):
        answer = answer_generation_chain.invoke({"query": q})["result"]

        print(f"Question {i}: {q}")
        print(f"Answer {i}: {answer}\n")

        f.write(f"Question: {q}\nAnswer: {answer}\n\n")
```
