## Advanced Pydantic

```python
from typing import List, Dict, Optional
from pydantic import BaseModel, Field
class PatientData(BaseModel):
    name: str
    age: int
    weight: float
    married: bool = False
    allergies: Optional[List[str]] = Field(default_factory=list)
    contact_info: Dict[str, str]
    
patient_data = {
    "name": "John Doe",
    "age": 30,
    "weight": 70.5,
    "married": "True",
    "allergies": ["peanuts", "shellfish"],
    "contact_info": {"email": "john.doe@example.com", "phone": "123-456-7890"}
}
patient_data_2 = {
    "name": "Alice Smith",
    "age": "20",
    "weight": 65.0,
    "contact_info": {"email": "alice.smith@example.com", "phone": "987-654-3210"}
}

patient_2 = PatientData(**patient_data_2)

patient = PatientData(**patient_data)

def display_patient_info(patient: PatientData):
    print(f"Name: {patient.name}")
    print(f"Age: {patient.age}")
    print(f"Weight: {patient.weight}")
    print(f"Married: {patient.married}")
    print(f"Allergies: {', '.join(patient.allergies)}")
    print(f"Contact Info: {', '.join(f'{k}: {v}' for k, v in patient.contact_info.items())}")
    
    print("-------------")

display_patient_info(patient)
display_patient_info(patient_2)
```

## Data Validation

```python
from typing import List, Dict, Optional, Annotated
from pydantic import BaseModel, Field, EmailStr, AnyUrl
class PatientData(BaseModel):
    name: str = Field( max_length=20, title="Name", description="Full name of the patient", example="John Doe")
    age: Annotated[int, Field(gt=0, lt=100, title="Age", description="Age of the patient", example=30)]
    weight: float = Field(strict=True, title="Weight", description="Weight of the patient", example=70.5)
    married: bool = False
    allergies: Optional[List[str]] = Field(default_factory=list)
    phone_no: Dict[str, str]
    email: EmailStr
    website: AnyUrl
    
patient_data = {
    "name": "John Doe",
    "age": 30,
    "weight": 70.5,
    "married": "True",
    "allergies": ["peanuts", "shellfish"],
    "phone_no": {"phone": "123-456-7890"},
    "email": "john.doe@example.com",
    "website": "http://johndoe.com"
}
patient_data_2 = {
    "name": "Alice Smith",
    "age": "20",
    "weight": 65.0,
    "phone_no": {"phone": "987-654-3210"},
    "email": "alice.smith@example.com",
    "website": "http://www.alicesmith.com"
}

patient_2 = PatientData(**patient_data_2)

patient = PatientData(**patient_data)

def display_patient_info(patient: PatientData):
    print(f"Name: {patient.name}")
    print(f"Age: {patient.age}")
    print(f"Weight: {patient.weight}")
    print(f"Married: {patient.married}")
    print(f"Allergies: {', '.join(patient.allergies)}")
    print(f"Phone No: {', '.join(f'{k}: {v}' for k, v in patient.phone_no.items())}")
    print(f"Email: {patient.email}")
    print(f"Website: {patient.website}")
    print("-------------")

display_patient_info(patient)
display_patient_info(patient_2)
```

## Field, Model and Computed Validator

```python
from typing import List, Dict, Optional, Annotated
from pydantic import BaseModel, Field, EmailStr, AnyUrl, field_validator, model_validator, computed_field
class PatientData(BaseModel):
    name: str = Field( max_length=20, title="Name", description="Full name of the patient", example="John Doe")
    age: Annotated[int, Field(gt=0, lt=100, title="Age", description="Age of the patient", example=30)]
    weight: float = Field(strict=True, title="Weight", description="Weight of the patient", example=70.5)
    height: float = Field(strict=True, title="Height", description="Height of the patient in meters", example=1.75)
    married: bool = False
    allergies: Optional[List[str]] = Field(default_factory=list)
    phone_no: Dict[str, str]
    email: EmailStr
    website: AnyUrl
    
    @field_validator("email", mode="after")
    @classmethod
    def validate_email(cls, value):
        valid_domains = ["icici.com", "sureify.com", "hdfc.com"]
        
        domain_name = value.split("@")[-1]
        if domain_name not in valid_domains:
            raise ValueError(f"Email must be from one of the valid domains: {', '.join(valid_domains)}")
        return value
    
    @field_validator("name", mode="after")
    @classmethod
    def transform_name(cls, value):
        return value.upper()
    
    @model_validator(mode="after")
    def validate_emergency_phone(cls, model):
        if model.age > 60 and "emergency" not in model.phone_no:
            raise ValueError("Patients over 60 must have an emergency contact")
        return model
    
    @computed_field
    @property
    def bmi(self) -> float:
        return self.weight / (self.height ** 2)  # Assuming age is used as a proxy for height in meters for BMI calculation
    
patient_data = {
    "name": "John Doe",
    "age": 30,
    "weight": 70.5,
    "height": 1.75,
    "married": True,
    "allergies": ["peanuts", "shellfish"],
    "phone_no": {"phone": "123-456-7890"},
    "email": "john.doe@sureify.com",
    "website": "http://johndoe.com"
}
patient_data_2 = {
    "name": "Alice Smith",
    "age": "20",
    "weight": 65.0,
    "height": 1.65,
    "phone_no": {"phone": "987-654-3210"},
    "email": "alice.smith@icici.com",
    "website": "http://www.alicesmith.com"
}

patient_2 = PatientData(**patient_data_2)

patient = PatientData(**patient_data)

def display_patient_info(patient: PatientData):
    print(f"Name: {patient.name}")
    print(f"Age: {patient.age}")
    print(f"Weight: {patient.weight}")
    print(f"Married: {patient.married}")
    print(f"Allergies: {', '.join(patient.allergies)}")
    print(f"Phone No: {', '.join(f'{k}: {v}' for k, v in patient.phone_no.items())}")
    print(f"Email: {patient.email}")
    print(f"Website: {patient.website}")
    print(f"BMI: {patient.bmi}")
    print("-------------")

display_patient_info(patient)
display_patient_info(patient_2)
```

## Nested Model and Serialization

```python
from pydantic import BaseModel

class Address(BaseModel):
    street: str
    city: str
    state: str
    zip_code: str
    
class Person(BaseModel):
    name: str
    age: int
    address: Address
    
address_1 = Address(
    street="123 Main St",
    city="Anytown",
    state="CA",
    zip_code="12345"
)
person_1 = Person(
    name="John Doe",
    age=30,
    address=address_1
)

address_2 = Address(
    street="456 Elm St",
    city="Othertown",
    state="NY",
    zip_code="67890"
)
person_2 = Person(
    name="Alice Smith",
    age=25,
    address=address_2
)

print(person_1)
print(person_2)
# Print the JSON representation of the person objects through the model_dump_json() method called Serializing to JSON
print(person_1.model_dump_json())
print(person_2.model_dump_json())
```
